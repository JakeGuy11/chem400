# Discussion 5: Zone Diagrams

We saw that, while all of our theories surround a single 1-electron transfer, multiple elementary reaction steps can chain together and alter the observed behavior. This field has been extensively studied and excellent numeric models exist for varied possible reaction sequences. This behavior can be broadly generalized as **E**lectron transfer steps or **C**hemical steps, with each being either "reversible" or "irreversible". The late Jean-Michel Saveant is largely responsible for pushing this field forward.

Several articles are listed below to provide you a sample of current research that involves advanced research on how multiple elementary reactions combine together to produce observed behavior.
