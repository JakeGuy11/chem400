# 2. Keeping a Handle on Your Figures

A figure you cannot refer to is a figure you cannot change. The habit below is worth forming now, because every figure in the rest of this course is built the same way: create the figure, create the axes, then talk to the axes by name.

````python
fig = plt.figure()      # define a pyplot figure named fig

ax = plt.subplot(111)   # 111 gives a single-panel figure

# Note that this command SPECIFICALLY calls the ax object, not the fig object
ax.plot(x, y1)           # pay attention to the format:   object.method(arguments)
````

````python
# create figure
fig2 = plt.figure(figsize=(7,2.625))   #we can define the size of the figure, inches by default, and the resolution in dpi
ax1 = plt.subplot(121)                      #1 row, 2 columns, index 1 (left panel)
ax2 = plt.subplot(122)                      #1 row, 2 columns, index 2 (right panel)

# Calculate two more data series
y2 = np.sin(2*x)                             # double the frequency
y3 = np.sin(3*x + np.pi/8)                  # triple it, and shift the phase

# sum of the three data series
ysum = y1 + y2 + y3

# plot out the data data
ax1.plot(x, y1, x, y2, x, y3)               # notice that we can plot multiple x,y pairs simultaneously. We could also do it across separate lines
ax2.plot(x,ysum)

# add labels
ax1.set_title('Individual Sine Waves')
ax2.set_title('An Interference Pattern')

for i in (ax1, ax2):                        # here we define a "tuple" and use it in a "for" loop to simplify a repetitive task. 
    i.set_xlabel('Time')
    i.set_ylabel('intensity')
````

One of the perks of assigning your figure to a variable is that it is now in memory, so you can add things to it on the fly, like change some formatting or write it to disk.

# 3. numpy Arrays

We used the np.linspace() command previously, but did not look at the format of the dataset that it generated. We can do that now:

````python
x = np.linspace(0, 1, 11)
print(f'x is a { type(x) }')
print(x)
````

As you can see above, the variable *x* refers to an object that is an <font color='blue'>numpy.ndarray</font>. You can think of this as being similar to a spreadsheet in software such as Microsoft Excel. There are conceptual differences in the indexing of items inside the two:
- Spreadsheets are usually indexed with letters and numbers, such as A1 or B10. 
- The content of numpy arrays are indexed by a comma-separated list of integers.

We can think of *x* as being a single column of data that can be accessed by an integer, *i*, denoting the row number:
```
x[i]
```

````python
x[3]
````

Rather than using the np.linspace() command, we can create these arrays directly:
```
variable_name = np.array()
```
Square brackets, as seen in the output from the second code cell above, are used to define your array.

````python
a_1d_array = np.array( [1,2,3,4] )
print(f'a_1d_array is a { type(a_1d_array) }')
print(a_1d_array)
print('Remember, indexing in python begins at 0!')
print(f'the second item is a { type(a_1d_array[1]) }')
````

If we want to create a multi-dimensional array, such as an (x,y) dataset, we can do so by increasing the number of brackets. Notice the format:
- Two sets of 4 numbers are contained in a first set of square brackets
- A second set of square brackets encapsulates those two sets

````python
# a_2d_array = np.array([[1,2,3,4],[5,6,7,8]])   #You CAN lay it out like this, but it is broken into multiple lines below to make the layout obvious.
a_2d_array = np.array( [
                       [1,2,3,4],
                       [5,6,7,8]
                       ] )

print(f'a_2d_array is a {type(a_2d_array)}')
print(a_2d_array)
print('Notice that we now need to provide two indices!')
print(f'the third item in the second set is a {a_2d_array[1,2]}')
````

Notice that I avoided the terms rows and columns. The data layout is arbitrary: you can think about it in any terms you like, and you can slice it in any way you like. We can access specific ranges inside the data by using colons:
```
array_item[i:j,m:n]
```
The line above will slice the data from index i to j in the first dimension, and m to n in the second dimension. Leaving any index blank will make the slice start at the beginning (i or m) or continue to the end (j or n). Try it out and pay attention to how the indices, and their ranges, are defined.

````python
print(a_2d_array[0,0:2])
print(a_2d_array[1,:])
print(a_2d_array[:,3])
print(a_2d_array[0:2,0:2])
````

The key concept here is that numpy arrays are infinitely scalable. We can adapt these array to any task we can imagine, but trying to manually define complex arrays rapidly becomes complicated.

# 4. Importing Real Data

With an understanding of how numpy arrays are formatted, we can now begin to import external data into our notebooks. We will do this using numpy's <font color="blue">genfromtxt()</font> method: 
```
np.genfromtxt(filename)
```
The input data must be a simple text file. It will be cut at every iteration of the *delimiter* character in the file. We will use a comma-separated text file that is structured into 
two columns with 11 rows as an example. You can create your own datafiles in MS Excel by using the "Save As..." feature and selecting CSV. The sample datafile we will load is located at:
```
../Resources/data/L3A.txt
```
You should navigate to this directory and double-click on the file to open it. Look at the layout of the data - it is exactly as you would see in a spreadsheet, with commas separating columns and new lines signifying new rows. Loading it into a numpy array is a matter of pointing to the file location:

````python
imported_data = np.genfromtxt('../Resources/data/L3A.txt', delimiter=',')
print(f'imported_data is a {type(imported_data)}')
print(f'array dimensions: {imported_data.shape}')
print(imported_data)
````

If we want to think in terms of spreadsheets, you can see that the first dimension in a numpy array corresponds to the ROW and the second to the COLUMN.

There are lots of optional arguments inside the np.genfromtxt() method. Place your cursor inside the genfromtxt() function in the cell above and press `shift+tab` to see a description of all of them. Some important ones are:
- delimiter=','
- skip_header = 0
- skip_footer = 0
 - change the 0 to the number of rows
- usecols = [comma separated list]
 - select a subset of columns to use

Note that `fmt` is **not** an argument of `genfromtxt()` - it belongs to `np.savetxt()`, which is the function that writes an array back out to disk. It is a string that sets the format of your numbers (i.e., decimal, exponential,...), such as `'%.3f'` or `'%5e'`. This will be important. A description of format strings can be found here: https://numpy.org/doc/stable/reference/generated/numpy.savetxt.html
 
Now that we have the data available to us, we can plot it:

````python
# create figure and axis
figL3A = plt.figure()
axL3A = plt.subplot(111)

# scatter plot with all rows from column 0 as x, all rows from column 1 as y
axL3A.scatter(imported_data[:,0], imported_data[:,1], marker='*')

# add some labels
# Latex
axL3A.set_title(r'A $\rightarrow$ B  with k = 5.0')  #the r'' enables the use of latex formatting in the title
axL3A.set_xlabel('time')
axL3A.set_ylabel('C$_A$ / (C$_A$ + C$_B$)')

figL3A.savefig('../Workspace/FigL3A.jpg')
````

Hopefully this graph looks familiar to you! This is an example of text book first-order decay kinetics.

Once loaded, we are free to handle the data in any way that we like. For example, perhaps we want to slice it into variables because...well its easier to remember a name than an index! We can then use these variables to do the next steps of the assignment:

````python
# define variables from the loaded data and use them to complete the exercise
time = np.copy( imported_data[:,0] )
relative_A = np.copy(imported_data[:,1])
relative_B = 1-relative_A
total_C = 0.035
absolute_A = relative_A*total_C
absolute_B = total_C-absolute_A

# create figure
figL3B = plt.figure(figsize=(7,2.625))
axL3B1 = plt.subplot(121)
axL3B2 = plt.subplot(122)

# subplot 1 is for normalized concentrations
axL3B1.plot(time, relative_A, marker='o', label=('C$_A$ / C$_A$ + C$_B$'))
axL3B1.plot(time, relative_B, marker='o', label=('C$_B$ / C$_A$ + C$_B$'))
axL3B1.set_xlabel('time')
axL3B1.set_ylabel('C$_i$ / (C$_A$ + C$_B$)')
axL3B1.legend()

# subplot 2 is for absolute concentrations
axL3B2.plot(time, absolute_A, marker='o', label=('C$_A$'))
axL3B2.plot(time, absolute_B, marker='o', label=('C$_B$'))
axL3B2.set_xlabel('time')
axL3B2.set_ylabel('C$_i$ (mol L$^{-1}$)')
axL3B2.legend()

figL3B.savefig('../Workspace/FigL3B.jpg')
````

The data you just plotted was generated from the equation for first order reaction kinetics:

\begin{equation}
C_A(t) = C_{A,0} \cdot e^{-kt}
\end{equation}

It was created in a notebook exactly like this one and written to disk with `np.savetxt()`. That is worth noticing: **the file you just imported was somebody's output.** Yours will be somebody else's input.

# Creating Custom Functions

One of the best things about using python for research is the ability to easily define custom workflows. Think about the flow of your research AFTER you finish acquiring your data. A function is very easy to create and offers infinite flexibility:

```
def function_name(var1, var2):
 code line 1
 code line 2
 return 
```

You simply use the <font color='blue'>def</font> command to create it, followed by a (meaningful) function name, a comma-separated list of desired input variables enclosed in round brackets, and a colon. All subsequent lines that are **indented by 4 spaces** become a part of this function. The first line that has no indentation formally ends the function.
- Inside the Jupyter environment we can cheat and use a `tab`, but that is because the notebook automatically turns the tab into 4 spaces (you can go into a code block, press tab, then press the left arrow to confirm this).

Here, we will make a function to iteratively generate data and add it to a plot. In future Lessons we will make use of functions to facilitate the visualization of data and structures.

The data displayed above for a first order reaction was generated using an arbitrary k value of 5.0. What if we wanted to visualize the data for various k values? A function makes this easy:

````python
def rxn_decay(x=None, k=1.0, a=1.0):
    """
    First order decay:  C(t) = a * exp(-k * t)

    The "doc-string" is a series of comments you can place at the start to
    describe the purpose of the function and the requirements for its variables.
    Notice that we provide default values, so we only have to pass the variables
    we actually want to change.

    Note also how the default for x is handled.  Writing
        def rxn_decay(x=np.linspace(0, 1, 2), ...)
    would build that array ONCE, when the function is defined, and then reuse
    the same object on every call.  For the mutable objects we met in Lesson 1
    that is a trap.  Using None and building the array inside the function
    avoids it entirely.
    """
    if x is None:
        x = np.linspace(0, 1, 2)
    y = a*np.exp(-k*x)
    return y
````

The <font color='blue'>return</font> command will give the user the listed object(s) upon completion of the function. you can return as many objects as you like, simply list them as comma-separated values.

We can now use the function to rapidly generate a plot for a given k value. Here, we'll do it using a finer spacing in our x-axis:

````python
t = np.linspace(0.,1.,50) #increase the number of steps in our x axis
k = 10.

y1 = rxn_decay(t, k)       #notice that we left out the 'a' variable. It will default to 1.0. 

plt.plot(t, y1, marker='o')
````

Our custom function facilitates our ability to calculate a dataseries using the first order reaction equation. It grants us control over the x-axis, the rate constant and the initial concentration. If we want to exploit this functionality to view multiple plots in the same figure, we can do so using the logical operators introduced in [Lesson1](L01-Introduction.ipynb). What we will do is define a series of k values for plotting. We will define them to be evenly spaced on an exponential scale, then plot kinetic traces for all of them on the same graph.

````python
# k values evenly spaced on exponential scale
k = 10**np.linspace(-1,1.5,20)

figL3C = plt.figure()
axL3C = plt.subplot(111)
axL3C.set_xlabel('time')
axL3C.set_ylabel('C$_i$ / C$_A$ + C$_B$')

for i in k:
    y = rxn_decay(t, i)
    axL3C.plot(t, y, label=str(i))

#figL3C.legend()
figL3C.savefig('../Workspace/FigL3C.jpg')
````

In practice, we can use functions to do many things - load data, process data, save data and figures, format figures, apply labels, format output for printing on screen, or anything else you can dream up.
