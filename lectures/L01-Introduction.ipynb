# String Objects
Lets tackle some simple text first. Text is stored as objects called <font color=blue>*strings*</font>. It is defined by equating a variable name with some text inside single or double quotations:
```

```

````python
my_string = "Hello class!"
````

There are numerous actions, or <font color=blue>*methods*</font>, that we can perform on these objects. Here are a few examples using strings.

Note that the variable defined in the cell above (my_string) is now stored in memory, so we can recall it in later cells:

````python
# The hashtag makes everything AFTER it invisible to Python. This is a COMMENT, which is a fantastic way to keep simple notes. Remember though, keep it clean!
print(my_string)                       # show the contents of the string   
print( len(my_string) )                # Give the length of the variable - here it is 10 letters, a space and a !
print(my_string[0:5])                  # splice the string

print(my_string[0:5] + my_string[-6:-1])  # CONCATENATE a string with a simple '+' operator. The numbers in the square brackets are integers that define the position index. 
                                      # Change the index values and get comfortable with how they work.
    
print(type(my_string))                 #see the OBJECT TYPE. Note that the print method here has a new string directly fed to it.

print('\n %s \n The object is a %s that has %s characters' %(my_string, type(my_string), len(my_string)))  #Here we do 3 things: 
                                      # use 'escape sequence' to add line breaks (\n), 
                                      # define a new string on the fly (my_string is stored in memory, the new string is not), 
                                      # and use a formatting command ( %s ) to insert two objects into the string. 
````

# Numeric Objects
Of course, as scientists, we want to focus on numbers rather than strings. Python comes with three basic number classes:
- <font color=blue>*int*</font> are integers
- <font color=blue>*float*</font> are floating point numbers
- <font color=blue>*complex*</font> are complex numbers

You can perform all basic operations the same way you would in Microsoft Excel (+ - * /). More advanced math operations can be done in simple ways:
- <font color=blue>*//*</font> divides, chops off decimal and provides you the integer
- <font color=blue>**</font> exponential
- <font color=blue>%</font> divides and provides you the *remainder*

Pay careful attention to how you define your numeric variables because it sets their <font color=blue>*type*</font>! In the examples below, notice that mixing int and float yields a response as a float. This minimizes chances of error in your mathematical operations, BUT some methods or functions require int as inputs and will crash if you try give it a float! If you are having errors in your code and don't understand them, check to make sure that your variables are the correct TYPE/CLASS.

````python
a = 9
b = 5.
print(f'a is {a}, a {type(a)}. \nb is {b}, a {type(b)} \n') # an 'f-string' inserts variables into strings

a = float(a)                                             # here, we redefine a variable as a modified version of the old one.
a = float(2)                                             # achieves similar outcome. A number without a decimal is defined by default as an int, but we can force it to be a float.
b = int(b)                                               # here, we redefine the type
print(f'a is now {a}, a  {type(a)} \nb is now {b}, a {type(b)} \n')

print('Demonstration of operations:')
print('%s/%s = %s' %(b, a, b/a))                         # an (older) alternative to f-strings
print('%s//%s = %s' %(b, a, b//a))
print('%s %s %s = %s' %(b, '%', a, b%a))
print('%s %s %s = %s' %(b, '**', a, b**a))
````

There exists an enormous number of operations you can perform on numeric objects, ranging from trivial to complex.

# Mutability and Identity
The concept of <font color="blue">*mutability*</font> can be confusing to newcomers to computer code. Mutability refers to the ability to change an object/variable after you create it. 

**Int, float and complex** numbers in python are immutable. Once defined, **they cannot be modified**! Changes require the redefinition of the object. You can see this in the cell below.
- The `id(variable_name)` command tells you where in the computer's memory that object is stored

````python
z = 1
print(id(z))
z = 2
print(id(z))
````

This leads to some interesting traps for the unwary! Watch what happens in the next cell:

````python
z = 1
x = z
print(f"First:\nz = {z} at {id(z)} \nx = {x} at {id(x)}")

z = 2
print(f"Now:\nz = {z} at {id(z)} \nx = {x} at {id(x)}")
````

We defined an int with `z = 1` then defined a second variable, x, which pointed at this first object. When we **redefined** z as a different integer we **did not change** the first object -- x still points at the original object and z now points at a new one!

This same situation happens below:

````python
z = 3
print(f"First:\nz = {z} at {id(z)}")
z = z + 1
print(f"Now:\nz = {z} at {id(z)}")
````

If you are ever in doubt, you can use the `is` command. This is NOT checking for equivalence in values, it is checking to see whether two variables refer to the same object. In short, it is checking equivalence of the object IDs:

````python
z = 1
x = z
print(z is x)

z = 2
print(z is x)
````

Strings are also immutable objects.

````python
z="Test "
print(z + str(id(z)))

z = z.replace('Test','Did we define a new object? ')
print(z+str(id(z)))
````

# Lists

A <font color=blue>*list*</font> is a mutable array of objects. You can make a list of as many different types of objects as you like - a list can even contain a second list nested inside of it. It is defined with a single set of **square brackets**, with objects separated by commas.
```
variable_name = [var1, var2, ...]
``` 

Lists are *indexed*, you can access entries with square brackets. **Careful, indexes start from 0**!
```
variable_name[int]
```

````python
my_list = [1, my_string, 'string defined on the fly']   # A list containing 3 objects

print(f"{my_list} is a {type(my_list)}")
print(f"{my_list[0]} is a {type(my_list[0])}")
````

### Objects with mutability MUST BE HANDLED WITH CARE! 

A frequent shortcut people try to take is to duplicate/copy one object by creating a new variable that points at an old one. However, if the original object is mutable, watch what happens...

````python
# define a new variable name that points to a MUTABLE list
my_new_list = my_list
print(my_list)
print(my_new_list)

# redefine the first object in the original list
my_list[0] = 'removed integer'

# check for equivalence
print(my_list is my_new_list)

# print both for inspection
print(my_list)
print(my_new_list)
````

We created a second variable that pointed at the original list. This seems like an intuitive way to create a new object, but you can see the problems that arise when we modify the original list. 
- We can see from the `is` command that the two variables point to the same object. 
- Unlike what we saw with immutable strings and numbers, we now changed the content of the original (mutable) list on-the-spot. 
- This can lead to extensive errors in calculations if you are not careful! If you want to duplicate or copy an object, the best choice is to formally copy/duplicate it:

```
variable.copy()
```

````python
my_new_list = my_list.copy()
my_list[1] = 10
print(my_list is my_new_list)
print(my_new_list)
````

Another way to copy a list is to iterate it by explicitly defining a new variable. The indexing of lists can be taken advantage of to make a new, more selective list.

````python
my_new_list = list(my_list)
print(my_new_list == my_list)  # check for content equivalence
print(my_new_list is my_list)  # check for identity equivalence

my_new_list = list(my_list[0:2])    # note that the index now has a : in it. This signifies the entire range between the two numbers. 
print(my_new_list)                  # note how the index counting works. It copied index 0 to 2, not including 2. (i.e. indexes 0 and 1 were copied)
````

It is very easy to manipulate the size and contents of a list.
- add *new_items* to *variable*
```
variable.append(new_items)
```


- add all elements from *list2* into *variable*
```
variable.extend(list2)
```


- add *new_item* into *variable* at the given index
```
variable.insert(index, new_item)
```


- removes the first occurrence of *item* from *variable*
```
variable.remove(item)
``` 


More methods exist, such as including sort(), reverse() and count(). 

# Tuples
A <font color=blue>*Tuple*</font> is a essentially an immutable form of a list. They are created using normal, rounded brackets.
```
variable_name = (var1, var2, ...)
``` 

Tuples are also *indexed*, you can access entries with square brackets. 
```
variable_name[int]
```

Lets try repeat all of the steps we used for lists:

````python
my_tuple = (1, my_string, 'defined on the fly')   #A tuple containing 3 objects

print(f"{my_tuple} is a {type(my_tuple)}")
print(f"{my_tuple[0]} is a {type(my_tuple[0])}")

my_new_tuple = my_tuple
print(my_tuple is my_new_tuple)
````

Now lets try redefine one of the objects:

````python
my_tuple[0] = 'removed integer'
````

**Note the error** - Python will not let you modify the immutable contents of a tuple. However, one small thing to be wary about is that while the tuple is immutable it may contain mutable objects such as lists. To see this in action, we will create a tuple that contains a list. We can then modify an item in the list and see the effect.

````python
t = (1, [1,2,3,4], 5)
print(t)
t[1][0] = 0  # note how we used double indices here
print(t)
````

Note that the first item in the internal list was changed! 
- The tuple is immutable. It will always point to the same 3 objects.
- The list, which is object 2 in the tuple, is mutable. We can continually change it without redefining where it is in the system's memory.
