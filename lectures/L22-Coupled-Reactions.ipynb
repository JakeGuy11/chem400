# 1. What happens when sequential chemical reactions occur?

Let us first consider the question of sequential reactions from a general chemistry perspective. A common example encountered throughout the chemistry curriculum is the dissolution of CO $_2$ in blood. The gas quickly reacts with H $_2$ O to create carbonic acid, which then deprotonates. The chain of reaction steps are:

$
\large CO_2 + H_2O \rightleftharpoons H_2CO_3
$

$
\large H_2CO_3 + H_2O \rightleftharpoons H_3O^+ + HCO_3^-
$

We can define thermodynamic equilibrium constants for each of these reaction steps. The *dissolution constant* ( $K_{d}$ ), which describes how much CO $_2$ will dissolve is:

$
\large K_{d} = \frac{P_{CO_2}}{C_{H_2CO_3}}
$

And the *acid dissociation constant* ( $K_a$ ) describes the acid-base equilibrium in solution:

$
\large K_a = \frac{C_{H_3O^+} \cdot C_{HCO_3^-}}{C_{H_2CO_3}}
$

The two reactions share a common intermediate and are therefore reliant upon one another. This interaction between elementary reactions can be captured by algebraically isolating the concentration of the shared species:

$
\large C_{H_2CO_3} = \frac{C_{H_3O^+} \cdot C_{HCO_3^-}}{K_a} = \frac{P_{CO_2}}{K_{d}}
$

We can then isolate a single term from this equivalence. In the current situation, it is convenient to isolate the concentration of protons, which is a variable we can experimentally measure:

$
\large C_{H_3O^+} = \frac{K_a \cdot P_{CO_2}}{K_{d} \cdot C_{HCO_3^-}} 
$

The two equilibrium **constants** can be combined ( $K = K_a /K_{d}$ ):

$
\large C_{H_3O^+} = K \cdot \frac{P_{CO_2}}{C_{HCO_3^-}} 
$

Taking the negative logarithm of each side, we obtain:

$
\large -log(C_{H_3O^+}) = -log(K) - log(\frac{P_{CO_2}}{C_{HCO_3^-}} )
$

Hopefully the alternative representation of this equation is obvious:

$
\large pH = pK + log(\frac{C_{HCO_3^-}}{P_{CO_2}})
$

Which can be readily compared to the fundamental equation used to describe Bronsted acid-base equilibria: the *Henderson-Hasselbalch equation*:

$
\large pH = pK_a + log(\frac{C_{A^-}}{C_{HA}})
$

# 2. The Electrochemistry-Specific Situation
Let us now consider the most basic situation in the context of electrochemistry: a proton-coupled electron transfer. These are very important and common reaction mechanisms because they are a mechanism that prevents localized charge accumulation.

<div class="alert alert-block alert-success">
<b>Proton-Coupled Electron Transfer (PCET)</b>: a reaction where a transfer of an electron is combined with the transfer of a proton. 
</div>

The generic overall reaction for a PCET is:

$
\large O + ne^- + mH^+ \rightleftharpoons RH_m^{(m-n)+}
$

For which an equilibrium constant ( $K_{eq}$ ) can be defined:

$
\large K_{eq} = \frac{a_{RH_m}}{a_O \cdot a_{H^+}^m}
$

This thermodynamic entity can be used within our thermodynamic equation for electron transfers:

$
\large E = E^{0} - \frac{RT}{nF} ln (Q) 
$

$
\large E = E^{0} - \frac{RT}{nF} ln (\frac{a_{RH_m}}{a_O \cdot a_{H^+}^m})
$

The standard potentials can be converted to formal potentials (see [Lesson 9](L09-Thermodynamics-I.ipynb)), and logarithmic terms can be separated, to better see the relationships:

$
\large E = E^{0'} + m \cdot \frac{RT}{nF} ln(C_{H^+}) - \frac{RT}{nF} ln\frac{C_{RH_m}}{C_{O}}
$

This form of the equation hopefully looks familiar - we saw a variation of it in [Lesson 14](L14-Mass-Transfer-Limitations.ipynb) where the $m_O/m_R$ ratio yielded a shift in $E_{1/2}$ . Based on similarities, we can deduce that the behavior for a PCET will be the same. Specifically, a change in [ $H^+$ ] will change observed $E_{1/2}$ values: 

$
\large E_{1/2} = E^{0'} +m\frac{RT}{nF}ln(C_{H^+})
$

<div class="alert alert-block alert-danger">
Thermodynamics is affected by the <b>overall</b> reaction that you are examining. If your experiment probes a reversible system (ie, thermodynamics dictates what you measure rather than kinetics), then any associated chemical reaction will alter the observed measured behavior. A PCET reaction is reliant upon the concentration of protons in your reaction medium!
</div>


Note that all of the effects we have discussed can interact with one another. For example, the $E_{1/2}$ in a PCET will be dependent on both the concentration of protons and relative changes in mass transfer coefficients. If we start with the generic equations for mass-transfer limited currents:

$
\large C_O(x=0) = \frac{i-i_{lim,c}}{nFAm_O}
$

$
\large C_{RH}(x=0) = \frac{i_{lim,a}-i}{nFAm_{RH}}
$

And substitute them into the Nernst equation:

$
\large E = E^{0'} +m\frac{RT}{nF}ln(C_{H^+}) - \frac{RT}{nF}ln\left[\frac{m_O}{m_R} \right] - \frac{RT}{nF}ln\left[\frac{C_{RH}}{C_O}\right]
$

The location of the redox process can therefore be described as:

$
\large E_{1/2} = E^{0'} +m\frac{RT}{nF}ln(C_{H^+}) - \frac{RT}{nF}ln\left[\frac{m_O}{m_R} \right]
$

or 

$
\large E_{1/2} = E^{0'} - \frac{m}{n}\cdot 0.059 \cdot pH - \frac{0.059}{n} \cdot log\left[\frac{m_O}{m_R} \right]
$

## Exploration of Expected Behavior 

We can design some functions and use them to explore the expected behavior:

```python
# echem.pcet_half_wave(pH, e_formal, m_over_n) applies
#
#     E_1/2 = E0' - (m/n) * 2.303 * (RT/F) * pH
#
# which is a slope of -59.2 (m/n) mV per pH unit at 298 K.
help(echem.pcet_half_wave)
```

We will start with the simplest situation, where a PCET is active across the full range that we measure. Lets imagine the reaction:

$
\large O + mH^+ + ne^- \rightleftharpoons RH_m^{(m-n)+}
$

The code box below allows us to explore this generic reaction, which will have some associated $E^0$ value:

```python
@interact(m=widgets.BoundedIntText(value=1, min=0, max=6, description=r'$m$'),
          n=widgets.BoundedIntText(value=1, min=1, max=6, description=r'$n$'),
          e0=widgets.FloatText(value=0.0, step=0.1, description=r'$E^0$'))
def plot_pourbaix_line(e0, m, n):
    '''One PCET, active across the whole pH range: a single sloping line.'''
    fig = plt.figure()
    ax = plt.subplot()
    ax.set_xlabel(r'pH')
    ax.set_ylabel(r'$E_{1/2}$ (V)')
    ax.set_xlim(0, 14)
    ax.set_ylim(-1.5, 1.0)

    e_half = echem.pcet_half_wave(PH, e_formal=e0, m_over_n=m/n)

    ax.plot(PH, e_half, 'o')
    ax.fill_between(PH, -10, e_half, color='blue', alpha=0.1)
    ax.fill_between(PH, 10, e_half, color='red', alpha=0.1)

    # Report the slope in mV/pH, keeping its sign.  np.polyfit gives both.
    slope_mV = 1000*np.polyfit(PH, e_half, 1)[0]
    ax.annotate(f'slope = {slope_mV:.1f} mV/pH', xy=(0.3, 0.9),
                xycoords='axes fraction')
    ax.annotate('O', xytext=(0.5, e0 + 0.3), xy=(0.5, e0 + 0.3),
                xycoords='data', color='red')
    ax.annotate('RH', xytext=(0.5, e0 - 0.5), xy=(0.5, e0 - 0.5),
                xycoords='data', color='blue')
```

Formally, we really should envision a reaction such as a PCET as a chain of **elementary steps**. Recall our discussion on the electrochemical potential in [Lesson 9](L09-Thermodynamics-I.ipynb) to understand why:

$
\large O + e^- \rightleftharpoons R^-
$

$
\large R^- + H^+ \rightleftharpoons RH
$

An electron transfer would create a negatively charged product; a prompt proton transfer from the surrounding medium would neutralize this charge. Thermodynamically, we therefore need to consider both $E^0$ and $pK_a$ . 

The overall (thermodynamic) picture can be generalized to the two extremes:

1. Solution pH is below the $pK_a$ of RH, such that the reduced product WILL be protonated:

$
\large O + H^+ + e^- \rightleftharpoons RH
$

2. Solution pH is above the $pK_a$ of RH such that the reduced product WON'T be protonated:

$
\large O + e^- \rightleftharpoons R^-
$

```python
@interact(pka=widgets.FloatText(value=7, description=r'$pK_a$'),
          e0=widgets.FloatText(value=0.0, step=0.1, description=r'$E^0$'))
def plot_pourbaix_with_pka(e0, pka):
    '''Above the pKa the proton is already gone, so the pH dependence stops.'''
    if not 0 < pka < 14:
        print('Choose a pKa between 0 and 14 so that both regimes are visible.')
        return
    fig = plt.figure()
    ax = plt.subplot()
    ax.set_xlabel(r'pH')
    ax.set_ylabel(r'$E_{1/2}$ (V)')
    ax.set_xlim(0, 14)
    ax.set_ylim(-1.5, 1.0)

    # Below the pKa: one proton per electron, so the line slopes
    low_ph = np.arange(0, pka, 0.3)
    e_half_low = echem.pcet_half_wave(low_ph, e_formal=e0, m_over_n=1.0)

    # Above the pKa: no proton transferred, so the line is flat.  It is offset so
    # that the two branches meet at the pKa.
    high_ph = np.arange(pka, 14.1, 0.3)
    e_half_high = (echem.pcet_half_wave(high_ph, e_formal=e0, m_over_n=0.0)
                   + echem.PCET_SLOPE*pka)

    pH = np.hstack((low_ph, high_ph))
    e_half = np.hstack((e_half_low, e_half_high))

    ax.plot(pH, e_half, 'o')
    ax.fill_between(low_ph, e_half_low, e_half_low - 10, color='blue', alpha=0.1)
    ax.fill_between(high_ph, e_half_high, e_half_high - 10, color='green', alpha=0.1)
    ax.fill_between(pH, e_half, e_half + 10, color='red', alpha=0.1)

    ax.annotate(r'$pK_a$', xy=(pka - 0.4, e_half.min() - 0.5), xycoords='data')
    ax.annotate('O', xy=(0.5, e0 + 0.3), xycoords='data', color='red')
    ax.annotate('RH', xy=(0.5, e0 - 0.5), xycoords='data', color='blue')
    ax.annotate(r'R$^-$', xy=(12., e_half.min() - 0.5), xycoords='data', color='green')
```

A more interesting situation arises when multiple protons can be transferred. Consider the overall reaction:

$
\large O + 2H^+ + e^- \rightleftharpoons RH_2^+
$

This reaction must be broken up into **three** elementary steps:

$
\large O + e^- \rightleftharpoons R^-
$

$
\large R^- + H^+ \rightleftharpoons RH
$

$
\large RH + H^+ \rightleftharpoons RH_2^+
$

The relative thermodynamics of each of these steps dictates the behavior that one would observe. Studying such multistep reactions is a major area of modern research. A relatively straightforward example can be seen in metal-organic compounds that are studied as *homogeneous electrocatalysts* for reactions such as the oxygen evolution reaction. Understanding the sequence of elementary steps enables researchers to establish a catalytic reaction mechanism cycle, which ideally enables improvement of catalytic performance based on application of rational chemical principles. # 3. The Influence of Chemical Reactions

It should be highlighted that the behavior exhibited in PCET reactions can be generalized to other reactions. Consider the following series of reactions that describe the formation of a complex between a metal ion ( $M^{n+}$ ) and ligands ( $X$ ), followed by an electron transfer:

$ 
\large M^{n+} + pX \rightleftharpoons MX_p^{n+}
$

$
\large MX_p^{n+} +ne^- \rightleftharpoons MX_p
$

The thermodynamics of the electron transfer step can be described by the Nernst equation:

$
\large E = E^{0'} -\frac{RT}{nF}ln\left[\frac{[MX_p]}{[MX_p^{n+}]} \right]
$

The complexation reaction has a thermodynamic formation constant, $K_f$ :

$
\large K_f = \frac{[MX_p^{n+}]}{[M^{n+}][X]^p}
$

Rearrangement yields:

$
\large [MX_p^{n+}] = K_f[M^{n+}][X]^p
$

Substitution into the Nernst equation:

$
\large E = E^{0'} + \frac{RT}{nF}ln[K_f] + \frac{p}{n}\frac{RT}{F}ln[X] -\frac{RT}{nF}ln\left[\frac{[MX_p]}{[M^{n+}]} \right]
$

It follows that:

$
\large E_{1/2} = E^{0'} + \frac{RT}{nF}ln[K_f] + \frac{p}{n}\frac{RT}{F}ln[X]
$

From this equation, it is clear that both thermodynamics and the mechanism can be studied by variation of the concentration of ligands in solution. This approach is referred to as the DeFord-Hume method.

```python
@interact(kf=widgets.FloatText(value=0.1, step=0.01, description=r'$K_f$'),
          p=widgets.IntSlider(value=1, min=1, max=3, description=r'$p$'),
          n=widgets.IntSlider(value=1, min=1, max=3, description=r'$n$'),
          e0=widgets.FloatText(value=0., step=0.01, description=r"$E^{0'}$"))
def plot_deford_hume(kf, p, n, e0):
    '''The DeFord-Hume method: E_1/2 against log[X] gives p from the slope
    and K_f from the intercept.  Same algebra as the Pourbaix diagram, with
    ligand concentration in place of pH.'''
    if kf <= 0:
        print('K_f must be greater than zero.')
        return
    ligand = np.logspace(-6, 1, 100)                # [X]
    slope = 2.303*p/(n*f)
    intercept = e0 + 2.303*np.log10(kf)/(f*n)
    e_half = intercept + slope*np.log10(ligand)

    fig = plt.figure()
    ax = plt.subplot()
    ax.set_xscale('log')
    ax.plot(ligand, e_half)
    ax.axvline(1, linestyle=':', color='gray')
    ax.annotate(f'slope={1000*slope:.1f} mV/decade, intercept={intercept:.3f} V',
                (ligand.min(), e_half.max()), fontsize=6)
    ax.set_xlabel('[X]')
    ax.set_ylabel(r'$E_{1/2}$ /V')
```
