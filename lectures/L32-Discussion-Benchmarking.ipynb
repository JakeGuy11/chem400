# Discussion 6: Benchmarking, Reproducibility and Reporting

Two papers report an oxygen evolution catalyst. One quotes an overpotential of 280 mV, the other 350 mV. Which is better? On the evidence given, **you cannot tell, and neither can the authors.** The numbers are only comparable if both groups normalised the current the same way, corrected for the same resistance, used the same electrolyte purity, and held the measurement for the same length of time. Very often they did not.

This is the last discussion of the course, and it is deliberately placed at the end. You now know enough to see where a number comes from, which means you now know enough to see where it can go wrong.
