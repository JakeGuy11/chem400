# 1. Bibliographic Tools 

## Software 
It is important to keep a record of what you're reading when you research a topic in the literature. Your process could be as simple as using something like Word or Excel to keep notes on the papers you read. However, there are a lot of modern tools that will make things easier for you.

Two very popular free options include: 

- [Mendeley](https://www.mendeley.com/?interaction_required=true)
- [Zotero](https://www.zotero.org/)

These packages are quite similar to each other, with the main difference being that Mendeley is offered by a commercial publisher while Zotero is a free and open-source software package developed by the community. Both have:
- Ability to manage a library of references (think iTunes for PDFs)
- Ability to mark up PDFs and take notes on them
- Add-ons that allow you to capture references and their PDFs direct from common web browsers
- Add-ons that allow you to insert citations and a bibliography directly into a Word document

The University of Waterloo library maintains some resources for [Reference Management Software](https://uwaterloo.ca/library/research-supports/reference-management-software)

## Citations

Your written report will require you to discuss the work of others. This means you have to cite it effectively. In practice, you add a *citation* in the document where relevant. This is generally a superscripted number at the end of the sentence. $^1$ In some formats, it could instead be a bracketed number [1]. The citations are all numbered in the order in which they appear, then listed in a reference list/bibliography at the end of the document. Completeness and formatting of the reference list are important. Read through ACS Style Guide for information on this:

[ACS Style Guide](https://pubs.acs.org/doi/full/10.1021/acsguide.40303)

# 2. Google Scholar

Google is exceptional at digging through an immense amount of information to give you what you searched for. This can be empowering, but also distracting. My recommendation is to use Google, or the specialized Google Scholar, to initiate a search when you know very little about what you're looking for. 

When you don't know what you're looking for, you need to gain exposure until you decide that you do know what you're looking for. You may decide you want to research electrocatalysis, but have no idea what terminology to use for searching. Google will present you with a wealth of information - but experience shows that this information is quite biased and prone to "Google bombing". If you solely rely on Google to search, you are likely to struggle to get very deep in the knowledge you seek.

# 3. Web Of Knowledge

Web of Knowledge is a search engine that is specialized for navigating the academic literature. It is far more selective than a Google search, which makes it challenging to get started with, but ultimately provides you an ability to ask better questions and find better materials. It is a subscription based service that UWaterloo pays for; you will need to be on campus (eduroam or hardwired), or connected to the VPN while at home.
