## Coverage

The midterm covers **Lessons 3 to 17**, which is the whole of the fundamentals half of the course:

- Foundations: the energy dilemma and the Ragone plot, the conceptual framework of an electrochemical experiment, cells and electrodes, resistance and the electrical double layer (Lessons 3, 4, 6, 7)
- Thermodynamics: electrochemical potential, the Nernst equation, reversibility, reference scales (Lessons 9, 10)
- Kinetics: potential energy surfaces, Butler-Volmer, exchange current, Tafel analysis, multi-step reactions (Lessons 11, 12, 13)
- Mass transport: diffusion, migration, convection, concentration profiles, the mass transfer coefficient (Lesson 14)
- Charge balance: liquid junctions, transference numbers, membrane potentials (Lesson 15)
- Techniques: chronoamperometry, Fick's laws and the Cottrell equation (Lesson 16)
- The material of Discussions 1 to 3 (Lessons 5, 8, 17)

**Lessons 1 and 2 are not examined** -- Jupyter and the Python language are tools for the course, not content to be tested.

Everything from Lesson 19 onwards is examined on the final. The fundamentals above remain assumed knowledge there.

## Format

Three sections, as in previous years:

1. **Compare and Differentiate.** Choose three of five pairs of terms. Explain how the members are related and how they differ. 4 marks each.
2. **Define and Describe.** Choose three of five terms. Give a detailed definition and describe their impact on electrochemical experiments. 4 marks each.
3. **Short Answers.** Answer two of three sets of questions. 12 marks each.

**Graduate students** additionally answer Section 4, a single synthesis question, and are marked against the standard set out in the syllabus: assumptions stated, units correct throughout, and at least one connection to experimental practice.
