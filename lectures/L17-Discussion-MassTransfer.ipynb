# Discussion 3: Mass Transfer

Electrochemical measurements always measure the AVERAGE behavior across an electrode. You should consider this each and every time you set up an experiment, or analyze experimental results. As we have started to see, mass transfer is one of the major phenomena that influence electrochemical behavior. The equations describing mass transfer that we will see in this class are all developed for **well defined electrode geometries**. Deviation from such geometry will result in deviation from the equations.

Several articles are listed below to provide you a sample of current research that involves various aspects of mass transfer. 
- **Read through the articles linked below before class.** 
- Highlight parts of the articles that you found informative and/or confusing.
