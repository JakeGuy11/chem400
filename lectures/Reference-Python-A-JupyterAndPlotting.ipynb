## Persistence of Objects
We cannot recall or modify any objects in later code cells unless we explicitly assign them to a variable. 

The figure above was plotted without assigning it to a variable; we have no means of finding it in the computer's memory. If we want to access it later we have to remake it. We can see this in action by trying to plot a second set of data, (x, y2), into the figure in a new cell:

````python
y2 = 2*np.sin(2*x-np.pi/4)
plt.plot(x, y2)
````

We DID save **x**, **limit** and **y1** into memory by assigning variables. They are now accessible in any cell in the notebook by simply typing their variable name, as we did for **x** above.

- Note that after running the code to define a variable in a cell we can even scroll upwards and use that variable in an earlier cell in the Jupyter Notebook. Be careful, though, because this can cause some nasty loops if you aren't careful!
 
Lets assign a variable to a figure such that we can modify it later on. We will also begin creating <font color="blue">matplotlib.pyplot.subplot</font> objects: 

```
axis_name = plt.subplot(###)
```

Where the ### contains 3 digits that correspond, in order, to:
- the number of rows
- the number of columns
- the index
 
These subplots allow us to define and access separate groups of axes such that we can make multiple panel figures, maintain control over where any specific dataset gets plotted, and enable advanced formatting.

````python
fig = plt.figure()      # define a pyplot figure named fig

ax = plt.subplot(111)   # 111 gives a single-panel figure

# Note that this command SPECIFICALLY calls the ax object, not the fig object
ax.plot(x, y1)           # pay attention to the format:   object.method(arguments)
````

We will now add a new sine-wave to the figure which has double the frequency.

**Run the next cell several times and observe what happens**

````python
y2 = np.sin(2*x)
ax.plot(x, y2)

# an alternative is to use gca, which stands for "get current axis". The line below chains together two functions/methods
# fig.gca().plot(x,y2)

ax.set_title('Sine Waves')
ax.set_xlabel('x axis')
ax.set_ylabel('y axis')

fig             # naming the figure object redisplays it. plt.show() would show nothing here: the inline backend closes figures at the end of the cell that made them.
````

**You should have noticed** that the new higher frequency curve changed color every time you executed the cell. This is because you kept adding additional lines to the fig object - the changing colors are because a new dataset is continually being added on top of the old one. 

**Run the next code cell several times and watch what happens.**

````python
fig = plt.figure()  # define a pyplot figure named fig
ax = plt.subplot(111)  # the 111 are  rows,columns, index    --- a 122 would be the second panel in a figure with 1 row and 2 columns
ax.plot(x,y1)

y2 = np.sin(2*x-np.pi/4)
ax.plot(x,y2)

ax.set_title('Sine Waves')
ax.set_xlabel('x axis')
ax.set_ylabel('y axis')
````

The figure is now stable, with no color changes, because the cell redefines the **fig** object every time the cell is executed.

The point of this exercise is that it is easy to get lost in your code. The concept here holds true for all objects you define and use, such as numbers and numeric arrays. If you want to perform a mathematical operation on objects, you need to be careful not to accidentally introduce errors by repeating it n-times!

Lets approach figure making a more conventional way, where we create and plot the figure all at once. A third sine wave will also be defined, then a fourth variable will be defined that is the sum of all three. We will then plot them side-by-side for comparison.

````python
# create figure
fig2 = plt.figure(figsize=(7,2.625))   #we can define the size of the figure, inches by default, and the resolution in dpi
ax1 = plt.subplot(121)                      #1 row, 2 columns, index 1 (left panel)
ax2 = plt.subplot(122)                      #1 row, 2 columns, index 2 (right panel)

# Calculate our third data series
y3 = np.sin(3*x + np.pi/8)

# sum of the three data series
ysum = y1 + y2 + y3

# plot out the data data
ax1.plot(x, y1, x, y2, x, y3)               # notice that we can plot multiple x,y pairs simultaneously. We could also do it across separate lines
ax2.plot(x,ysum)

# add labels
ax1.set_title('Individual Sine Waves')
ax2.set_title('An Interference Pattern')

for i in (ax1, ax2):                        # here we define a "tuple" and use it in a "for" loop to simplify a repetitive task. 
    i.set_xlabel('Time')
    i.set_ylabel('intensity')
````

# Streamlining Your Code

As you can see, there is absolute flexibility in what you can do, and therefore in what you can accomplish. To avoid getting lost you need to think carefully about WHAT YOU WANT TO DO, then find the shortest way there. It is important to do occasional tests as you build up complex pieces of coding to ensure that it is behaving as you expect it to - think of it as calibration of your experiments.

The goal is to accomplish your task using the shortest, cleanest, most easily updatable way possible. The previous figure can be made from scratch like this:

````python
#Define our figure and subplot axes
fig3 = plt.figure(figsize=(7,2.625))
ax1 = plt.subplot(121)
ax2 = plt.subplot(122)
limit = 6*np.pi

#define values for sine wave frequency and phase shift in a numpy array, define custom line styles in a tuple, and labels for a legend in a tuple.
sinParameters = np.array(
    [
        [1., 4., 6.],
     [0.,np.pi/4,np.pi/8]
    ]
)
lineFormats = ('-','--','-.')
dataLabels = ('component 1','component 2','component 3')

#define x and initialize ysum 
x = np.linspace(0,limit,600)
ysum = x*0

#Calculate three different sine waves according to the parameters in the sinParameters array, then plot them to ax1
for i in range(sinParameters[0].size):                           
    y = np.sin(x*sinParameters[0,i]-sinParameters[1,i])
    ax1.plot(x, y, lineFormats[i], label=dataLabels[i])
    ysum = ysum + y

#Plot the sum of all individual waves in ax2    
ax2.plot(x,ysum)

#apply formatting to both sets of axes
for i in (ax1, ax2):               
    i.set_xlabel('Time')
    i.set_ylabel('Intensity')
    i.set_ylim(-3,2.5) #shared axis limit for both figure panels so we visualize differences in intensity
    i.set_xlim(0,limit)

#Place the legend in ax1 and show plot
ax1.legend(loc='upper right')
fig3.savefig('../Workspace/Lesson2-Figure2.jpg')
````

**Note:** A new (optional) feature will be introduced here - the use of the <font color='blue'>jupyter-matplotlib widget</font>. This widget is activated by what is known as a jupyter *magic*...which essentially means a shortcut. The first line of the following code block is this *magic*:

````python
%matplotlib widget
#Define our figure and subplot axes
fig4 = plt.figure(figsize=(7,2.625))
ax1 = plt.subplot(121)
ax2 = plt.subplot(122)

#define values for sine wave frequency and phase shift in a numpy array, define custom line styles in a "tuple" (a list), and labels for a legend in a tuple.
sinParameters = np.array([[1.,2.,3.],
                          [0.,np.pi/4,np.pi/8],
                          [-0.2, -0.1, -0.5]])
lineFormats = ('-','--','-.')
dataLabels = ('line1','line2','line3')

#define x and initialize ysum
limit = 6*np.pi
x = np.linspace(0,limit,600)
ysum = x*0     

#Calculate three different sine waves according to the data above, then add them to ax1
for i in range(sinParameters[0].size):                           
    y = np.power(np.e,x*sinParameters[2,i]) * np.sin(2*x*sinParameters[0,i]-sinParameters[1,i]*np.pi)
    ax1.plot(x,y, lineFormats[i], label=dataLabels[i])
    ysum = ysum + y

#Plot the sum of all individual waves in ax2    
ax2.plot(x,ysum)

#apply formatting to both sets of axes
for i in (ax1, ax2):               
    i.set_xlabel('k')
    i.set_ylabel(r'$\chi$')
    i.set_ylim(1.1*ysum.min(),1.1*ysum.max()) #shared axis limit so we visualize differences in intensity
    i.set_xlim(0,limit)

#Place the legend in ax1 and show plot
ax1.legend(loc='upper right')
fig4.savefig('../Workspace/Lesson2-Figure3.png')
````
