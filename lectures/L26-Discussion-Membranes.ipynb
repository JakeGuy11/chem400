# Discussion 4: Membranes

Conceptually, a membrane is included in an electrochemical cell simply to restrict certain species from travelling to where you don't want them. Practically, this becomes quite challenging. We must ensure that ion flow between the electrodes is not disrupted; any modification to the flow of solvated species therefore has a thermodynamic effect.
