## 2. Course Description:

This course will introduce the fundamental concepts that underlie modern electrochemical energy conversion and storage technologies. These concepts will be used to build an understanding of common experimental techniques, such as cyclic voltammetry, chronoamperometry and impedance spectroscopy and their application to the analysis of real-world electrochemical systems. Modern technologies for electrochemical energy conversion and storage will then be compared and contrasted through a series of literature-based case studies, with focus on the design of electrode materials and measurement of their electrochemical properties.

Course material will be delivered using Jupyter notebooks and snippets of interactive python code that participants can adapt to their own research. The course will begin with an introduction to the working environment and the use of python within it. The course content will then be delivered explored with the use of open-source python packages.

**Pre-requisite: Chemical thermodynamics and kinetics. No prior knowledge of the python language or programming is necessary.**
 

### Topics:

The content and order of content delivery are subject to change:
- Introduction to Jupyter Notebooks and the python language
- thermodynamics in electrochemical systems
- kinetics in electrochemical systems
- mass transport in electrochemical systems
- a selection of electrochemical techniques 
- chemical information that can be experimentally extracted
- technologies for electrochemical energy conversion and storage will be introduced
- research strategies and design principles for electrode materials will be analyzed.

### Learning Objectives:

Upon completion of the course, students will be able to:
- Describe the fundamental chemical concepts underlying the study of electron transfer reactions
- Identify experimental techniques to extract information regarding electron transfer reactions
- Predict current-voltage behavior for model electron transfer systems
- Infer chemical processes underlying a current-voltage response
- Differentiate between electrochemical energy storage technologies
- Critically appraise strategies to improve electrochemical energy storage technologies

### Artificial Intelligence
This course exists so that you develop specific skills and judgment. Generative AI (GenAI) tools are now part of ordinary scientific workflow. You are welcome to use them for the research paper and the presentation components of this course. They have no role in the midterm or final examination and are expressly forbidden for them. Two conditions on the use of GenAI tools apply:

**Disclose.** Submit a short appendix stating which tools you used, what you asked them for, what you accepted, and what you rejected and why. Disclosure carries no penalty. Undisclosed use is an academic integrity matter under Policy 71.

**Verify.** Every number, equation and citation in your report is your responsibility. GenAI tools invent plausible references and plausible-looking values. This is the same problem Discussion 6 raises about the literature: a number means nothing until you know how it was produced. Include a verification table listing each citation and each quantitative claim with the primary source you checked it against.

All students are fully responsible for the work they submit, which means that you are expected to be capable of explaining every aspect of it. You should expect a brief conversation about your submitted work. In this conversation you will be asked to explain a figure, defend a claim, or clarify a statement. This applies to everyone and is not an accusation.
